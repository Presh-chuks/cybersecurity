```python
import math
from math import factorial, log10, pi, sqrt, prod
 
 # Factorial example 1 (find factorial 4)
a = 4 
print(factorial(a))

#exaple 2 (simplify)
b = 8
c = 5
factorial(b) / factorial(c)

#log10 example 1 
log10(1000)

# example 2 (log10 using power rule)
num = 3**2
print (log10(num))

# PI Example 1 (find the area of a circle with radius 5)
p = pi * 5**2
print(p)

# find the circumference of radius = 7cm
Circum = 2 * pi * 7
print(Circum)

#Sqrt Example 1 
num2 = sqrt(72)
print(num2)


#Product example 1
numbers = [5,7,8,9,21,30]
product = prod(numbers)
print(product)
 
#example 2 
num3 = (3*2)
num4 = (4*5)
sum = num3 * num4
print(sum)

```
